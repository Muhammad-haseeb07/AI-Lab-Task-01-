```python

graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": ["G"],
    "G": []
}

def dls(node, goal, limit, depth, path, visited, cutoffs):
    visited.append((node, depth))

    if node == goal:
        return "FOUND", path

    if depth == limit:
        for child in graph[node]:
            if child not in path:
                cutoffs.append(node)
                return "CUTOFF", None
        return "FAILURE", None

    cutoff_found = False

    for child in graph[node]:
        if child not in path:
            status, result = dls(
                child, goal, limit, depth + 1,
                path + [child], visited, cutoffs
            )

            if status == "FOUND":
                return status, result

            if status == "CUTOFF":
                cutoff_found = True

    if cutoff_found:
        return "CUTOFF", None

    return "FAILURE", None


# Test DLS at each depth limit
for limit in range(1, 5):
    visited = []
    cutoffs = []

    status, path = dls(
        "A", "G", limit, 0, ["A"], visited, cutoffs
    )

    print("\nDLS Limit:", limit)
    print("Status:", status)
    print("Path:", path if path else "No goal path")
    print("Visited:", visited)
    print("Cutoff nodes:", cutoffs)


# Iterative Deepening Search (IDS)
print("\n--- IDS RESULTS ---")

for max_depth in range(1, 5):
    print("\nMaximum depth:", max_depth)

    for limit in range(1, max_depth + 1):
        visited = []
        cutoffs = []

        status, path = dls(
            "A", "G", limit, 0, ["A"], visited, cutoffs
        )

        print("Limit:", limit, "| Status:", status)

        if status == "FOUND":
            print("Goal path:", path)
            break
```

```python

graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [], "E": [], "F": ["G"], "G": []
}

def dls(node, goal, limit, path):
    if node == goal:
        return "FOUND", path

    if limit == 0:
        return ("CUTOFF", None) if graph[node] else ("FAILURE", None)

    cutoff = False

    for child in graph[node]:
        if child not in path:
            status, result = dls(child, goal, limit - 1, path + [child])

            if status == "FOUND":
                return status, result
            if status == "CUTOFF":
                cutoff = True

    return ("CUTOFF", None) if cutoff else ("FAILURE", None)


for limit in range(1, 5):
    status, path = dls("A", "G", limit, ["A"])
    print("Limit:", limit, "| Status:", status, "| Path:", path)
```

```python
graph = {
    "A": [("B", 2), ("C", 3)],
    "B": [("D", 3), ("E", 2)],
    "C": [("F", 1)],
    "D": [],
    "E": [],
    "F": [("G", 0)],
    "G": []
}

h = {
    "A": 5,
    "B": 4,
    "C": 3,
    "D": 2,
    "E": 2,
    "F": 1,
    "G": 0
}

Visited = []
Queue = []

def Astar(node):
    Queue.append((h[node], 0, node))

    while Queue:
        Queue.sort()
        f, cost, temp = Queue.pop(0)

        if temp in Visited:
            continue

        Visited.append(temp)
        print("Current Node:", temp, "Cost:", cost, "F:", f)

        for child, n_cost in graph[temp]:
            if child not in Visited:
                new_cost = cost + n_cost
                new_f = new_cost + h[child]
                Queue.append((new_f, new_cost, child))

Astar("A")

print("Queue", Queue)
print("Visited", Visited)
```

```python

graph = {
    "A": [("B", 2), ("C", 3)],
    "B": [("D", 3), ("E", 2)],
    "C": [("F", 1)],
    "D": [],
    "E": [],
    "F": [("G", 0)],
    "G": []
}

Visited = []
Queue = []

def Ucs(node):
    Queue.append((0, node))

    while Queue:
        Queue.sort()
        cost, temp = Queue.pop(0)
        
        Visited.append(temp)
        print("Current Node:", temp, "Cost:", cost)

        for child, n_cost in graph[temp]:
            if child not in Visited:
                Queue.append((cost + n_cost, child))

Ucs("A")

print("Queue", Queue)
print("Visited", Visited)
```

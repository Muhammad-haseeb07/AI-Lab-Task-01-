```python
class ReflexACAgent:
    def __init__(self):
        self.previous_mode = None

    def choose_mode(self, temperature, desired, occupied):

        if occupied == False:
            target = "OFF"
        elif temperature > desired:
            target = "ON"
        else:
            target = "OFF"
            
        if target == self.previous_mode:
            command = "No command"
        else:
            command = "Send command"

        self.previous_mode = target

        return target, command
agent = ReflexACAgent()
tests = [
    (31, 30, True),
    (30, 30, True),
    (27, 28, True),
    (33, 30, False)
]
for temp, desired, occupied in tests:
    previous = agent.previous_mode

    target, command = agent.choose_mode(
        temp, desired, occupied
    )

    print(previous, temp, desired, occupied, target, command)


```

```python
graph = {
         "A":["B","C"],
         "B":["D","E"],
         "C":["F"],
         "D":[],
         "E":[],
         "F":["G"],
         "G":[],
}
visited = []
stack = []
def dfs(node):
    stack.append(node)
    while stack:
      print(stack,visited)
      temp = stack.pop()
      visited.append(temp)
      for child in graph [temp]:
        stack.append(child)
        

dfs("A")

print(stack,visited)
```

```python
graph = {
         "A":["B","C"],
         "B":["D","E"],
         "C":["F"],
         "D":[],
         "E":[],
         "F":["G"],
         "G":[],
}
visited = []
Queue = []
def bfs(node):
    Queue.append(node)
    while Queue:
      print(Queue,visited)
      temp = Queue.pop(0)
      visited.append(temp)
      for child in graph [temp]:
        Queue.append(child)
        

bfs("A")

print(Queue,visited)
```

```python
graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [], "E": [], "F": ["G"], "G": []
}

def dls(node, goal, limit, path):
    if node == goal:
        return "FOUND", path

    if limit == 0:
        return ("CUTOFF", None) if graph[node] else ("FAILURE", None)

    cutoff = False

    for child in graph[node]:
        if child not in path:
            status, result = dls(child, goal, limit - 1, path + [child])

            if status == "FOUND":
                return status, result
            if status == "CUTOFF":
                cutoff = True

    return ("CUTOFF", None) if cutoff else ("FAILURE", None)


for limit in range(1, 5):
    status, path = dls("A", "G", limit, ["A"])
    print("Limit:", limit, "| Status:", status, "| Path:", path)
```
